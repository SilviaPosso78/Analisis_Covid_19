1. Obtén un perfilado de datos para el DataFrame (20 puntos).

```python
# PERFILADO DEL DATAFRAME

# 1. Dimensiones del DataFrame
print("DIMENSIONES DEL DATAFRAME")
print(f"Filas: {df.shape[0]}")
print(f"Columnas: {df.shape[1]}")

# 2. Información general
print("\nINFORMACIÓN GENERAL")
df.info()

# 3. Tipos de datos
print("\nTIPOS DE DATOS")
display(df.dtypes.to_frame(name="Tipo de dato"))

# 4. Valores faltantes
print("\nVALORES FALTANTES")
faltantes = pd.DataFrame({
    "Valores faltantes": df.isnull().sum(),
    "Porcentaje faltante": (df.isnull().sum() / len(df) * 100).round(2)
})

display(faltantes.sort_values("Valores faltantes", ascending=False))

# 5. Valores únicos por columna
print("\nVALORES ÚNICOS")
unicos = pd.DataFrame({
    "Valores únicos": df.nunique(dropna=False)
})

display(unicos.sort_values("Valores únicos"))

# 6. Registros duplicados
print("\nREGISTROS DUPLICADOS")
print(f"Cantidad de registros duplicados: {df.duplicated().sum()}")

# 7. Estadísticas descriptivas de variables numéricas
print("\nESTADÍSTICAS DESCRIPTIVAS - VARIABLES NUMÉRICAS")
display(df.describe().T)

# 8. Estadísticas de variables categóricas
print("\nESTADÍSTICAS DESCRIPTIVAS - VARIABLES CATEGÓRICAS")
display(df.describe(include="object").T)

# 9. Ejemplo de frecuencias de las variables categóricas
print("\nFRECUENCIAS DE VARIABLES CATEGÓRICAS")

for columna in df.select_dtypes(include="object").columns:
    print(f"\n--- {columna} ---")
    display(df[columna].value_counts(dropna=False).head(10))
```

```python
# Clasificación de variables según su tipo

numericas = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categoricas = df.select_dtypes(include=["object", "category"]).columns.tolist()
fechas = df.select_dtypes(include=["datetime64[ns]"]).columns.tolist()

print("VARIABLES NUMÉRICAS:")
print(numericas)

print("\nVARIABLES CATEGÓRICAS:")
print(categoricas)

print("\nVARIABLES DE FECHA:")
print(fechas)
```
