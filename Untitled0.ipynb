### Identificación de registros duplicados

Para identificar registros duplicados se utilizó la función `duplicated()` de Pandas. Esta función permite detectar las filas que contienen exactamente la misma información que un registro previamente registrado. Con el resultado abtenido podemos determinar la cantidad y proporción de registros duplicados en el DataFrame.

En esta etapa los registros duplicados fueron únicamente identificados y no eliminados, debido a que se solicita su identificación y no la eliminación por ahora. Además, antes de eliminarlos es necesario determinar si corresponden realmente a registros repetidos o si representan casos válidos registrados más de una vez por criterios propios del proceso en la recopilación de información.

### Identificación de valores atípicos

Para identificar valores atípicos se utilizó la variable `Edad`, debido a que es una variable cuantitativa sobre la cual tiene sentido aplicar un análisis estadístico de valores extremos. Se utilizó el método del rango intercuartílico (IQR), calculando el primer cuartil (Q1), el tercer cuartil (Q3) y la diferencia entre ambos.

Se consideraron como posibles valores atípicos aquellos registros que se encuentran por debajo del límite inferior o por encima del límite superior, calculados mediante 1,5 veces el rango intercuartílico.

Es importante aclarar que un valor identificado como atípico desde el punto de vista estadístico no necesariamente representa un error en los datos. Por ejemplo, una edad elevada puede ser poco frecuente dentro del conjunto de datos, pero continuar siendo un valor válido. Por esta razón, los valores atípicos deben ser analizados antes de decidir si deben corregirse o eliminarse.

Formulas para obtener los datos

Diseña una fórmula para calcular la tasa de letalidad del Covid-19 en Colombia, Antioquia y Medellín. (10 puntos).

```python
def calcular_letalidad(df_subconjunto):
    """Calcula la tasa de letalidad (%) en un DataFrame o filtro específico."""
    total_casos = len(df_subconjunto)
    if total_casos == 0:
        return 0.0

    total_fallecidos = (
        (df_subconjunto['Estado'] == 'Fallecido')
        | (df_subconjunto['Ubicación del caso'] == 'Fallecido')
        | (df_subconjunto['Recuperado'] == 'Fallecido')
    ).sum()

    return (total_fallecidos / total_casos) * 100


# Cálculo para los tres niveles geográficos
letalidad_colombia = calcular_letalidad(df_copia)
letalidad_antioquia = calcular_letalidad(
    df_copia[df_copia['Nombre departamento'] == 'ANTIOQUIA']
)
letalidad_medellin = calcular_letalidad(
    df_copia[df_copia['Nombre municipio'] == 'MEDELLIN']
)

print(f"Tasa de Letalidad en Colombia: {letalidad_colombia:.2f}%")
print(f"Tasa de Letalidad en Antioquia: {letalidad_antioquia:.2f}%")
print(f"Tasa de Letalidad en Medellín:  {letalidad_medellin:.2f}%")
```

¿Cuáles son los retrasos promedio entre las fechas? (5 puntos).

```python
# 1. Retraso en el diagnóstico (Inicio de síntomas -> Diagnóstico)
retraso_diagnostico = (
    df_copia['Fecha de diagnóstico'] - df_copia['Fecha de inicio de síntomas']
).dt.days.mean()

# 2. Retraso en la notificación (Inicio de síntomas -> Notificación)
retraso_notificacion = (
    df_copia['Fecha de notificación'] - df_copia['Fecha de inicio de síntomas']
).dt.days.mean()

# 3. Tiempo promedio de reporte web (Diagnóstico -> Reporte web)
retraso_reporte = (
    df_copia['fecha reporte web'] - df_copia['Fecha de diagnóstico']
).dt.days.mean()

# 4. Tiempo promedio a la recuperación (Inicio de síntomas -> Recuperación)
tiempo_recuperacion = (
    df_copia['Fecha de recuperación'] - df_copia['Fecha de inicio de síntomas']
).dt.days.mean()

# 5. Tiempo promedio al deceso (Inicio de síntomas -> Muerte)
tiempo_deceso = (
    df_copia['Fecha de muerte'] - df_copia['Fecha de inicio de síntomas']
).dt.days.mean()

print(
    f"Retraso promedio Síntomas -> Diagnóstico:  {retraso_diagnostico:.1f} días"
)
print(
    f"Retraso promedio Síntomas -> Notificación: {retraso_notificacion:.1f} días"
)
print(
    f"Retraso promedio Diagnóstico -> Reporte Web: {retraso_reporte:.1f} días"
)
print(
    f"Tiempo promedio Síntomas -> Recuperación: {tiempo_recuperacion:.1f} días"
)
print(f"Tiempo promedio Síntomas -> Deceso:      {tiempo_deceso:.1f} días")
```
