2. Realiza una limpieza de datos de las variables anteriores, haciendo uso de .zip(), .map(), .replace(), .fillna(), etc. (10 puntos).



```python
import pandas as pd
import numpy as np

# 1 Variables solicitdas y columnas correspondientes

variables = {
    'Estado': 'Estado',
    'Atención': 'Ubicación del caso',
    'Sexo': 'Sexo',
    'Tipo': 'Tipo de contagio',
    'País_procedencia': 'Nombre del país',
    'Tipo_recuperación': 'Tipo de recuperación',
    'Pertenencia_étnica': 'Pertenencia étnica',
    'Nombre_grupo_étnico': 'Nombre del grupo étnico'
}


#2 Crear una copia para mantener el DataFrame original, y hacer lo cambios
df_limpio =df.copy()

#3. Reemplzar valores vacios o inconsistencias por NaN
valores_invalidos = ["", " ","NA", "N/A"]

for nombre, columna in variables.items():
    df_limpio[columna] = df_limpio[columna].replace(valores_invalidos, np.nan)

#4 Estandarización de la variable Sexo usando .map()
mapa_sexo = {
    'm': 'M',
    'f': 'F',
    'F': 'F',
    'M': 'M'
}

df_limpio['Sexo'] = (
    df_limpio['Sexo']
    .astype('string')
    .str.strip()
    .map(mapa_sexo)
)

#5. Estandarización de la variable Atención utilizando .replace()
mapa_atencion = {
    'casa': 'Casa',
    'CASA': 'Casa',
    'fallecido': 'Fallecido',
    'FALLECIDO': 'Fallecido',
    'hospital': 'Hospital',
    'HOSPITAL': 'Hospital',
    'hospital uci': 'Hospital UCI',
    'HOSPITAL UCI': 'Hospital UCI'

}

df_limpio['Ubicación del caso'] = (
    df_limpio['Ubicación del caso']
    .astype('string')
    .str.strip()
    .replace(mapa_atencion)
)

#6. Completar los valores faltantes de las variables
columnas_categoricas = [
    'Estado',
    'Ubicación del caso',
    'Sexo',
    'Tipo de contagio',
    'Nombre del país',
    'Tipo de recuperación',
    'Nombre del grupo étnico'
]

for columna in columnas_categoricas:
    df_limpio[columna] = df_limpio[columna].fillna(
        'Sin registrar'
    )

# 7. Completar los valores faltantes de Pertenencia étnica
df_limpio['Pertenencia étnica'] = (
    df_limpio['Pertenencia étnica'].fillna(-1)
)

# 8. Verificar el resultado utilizando zip()
print("=== VERIFICACIÓN DESPÚES DE LA LIMPIEZA===")

for nombre, columna in zip(variables.keys(), variables.values()):
    nulos= df_limpio[columna].isnull().sum()
    print(f"{nombre} - Valores nulos: {nulos}")

```

**Justificación**

Despúes de aplicar el proceso de limpieza,se realizó una verificación mediante ***.isnull().sum()***, para comprobar la presencia de valores nulos en las variables. El resultado indica que los valores faltantes identificados durante el proceso fueron tratados en las transfirmaciones realizadas.

En las variables categóricas, los valores faltantes fueron reemplzadas por "Sin registrar", y las variables númericas como **Pertenencia étnica** por ***-1*** ya que se requiere conservar el tipo de dato de la variable, que en este caso es númerico.Por lo tanto, los valores que inicialmente podían aparecer como NaN ya no son identificados por Pandas como valores nulos.

Con esta verificación se puede comprobar que el proceso de limpieza se ejecutó correctamente y que las variables se encuentran listas para continuar con el análisis.

**Justificación**

La decodificación de las variables de fecha se realizó correctamente utilizando **pd.to_datetime()**, convirtiendo las seis variables seleccionadas del tipo **object** a **datetime64[ns]**. Esto permite que Pandas reconozca las fechas como datos temporales y facilita operaciones posteriores, como calcular diferencias entre fechas, obtener años o meses y analizar períodos de tiempo.

Durante la conversión se identificaron valores **NaT**, principalmente en **Fecha de muerte**, **Fecha de inicio de síntomas** y **Fecha de recuperación**. Estos valores representan fechas ausentes y no necesariamente errores en los datos, ya que algunas personas no fallecieron o no cuentan con una fecha registrada. Diseña una fórmula para calcular la tasa de letalidad del Covid-19 en Colombia, Antioquia y Medellín. (10 puntos).

Formulas para obtener los datos

***Letalidad= número de fallecidos / Número toal de casos confirmados * 100***

```python
def calcular_letalidad(df_subconjunto):
    """Calcula la tasa de letalidad (%) en un DataFrame o filtro específico."""
    total_casos = len(df_subconjunto)
    if total_casos == 0:
        return 0.0

    total_fallecidos = (
        (df_subconjunto['Estado'] == 'Fallecido')
        | (df_subconjunto['Ubicación del caso'] == 'Fallecido')
        | (df_subconjunto['Recuperado'] == 'Fallecido')
    ).sum()

    return (total_fallecidos / total_casos) * 100


# Cálculo para los tres niveles geográficos
letalidad_colombia = calcular_letalidad(df_copia)
letalidad_antioquia = calcular_letalidad(
    df_copia[df_copia['Nombre departamento'] == 'ANTIOQUIA']
)
letalidad_medellin = calcular_letalidad(
    df_copia[df_copia['Nombre municipio'] == 'MEDELLIN']
)

print(f"Tasa de Letalidad en Colombia: {letalidad_colombia:.2f}%")
print(f"Tasa de Letalidad en Antioquia: {letalidad_antioquia:.2f}%")
print(f"Tasa de Letalidad en Medellín:  {letalidad_medellin:.2f}%")
```

**Justificación**

La tasa de letalidad se calculó mediante la relación entre el número de fallecidos y el número total de casos confirmados, multiplicada por 100. Los resultados obtenidos fueron de **2,24 % para Colombia, 1,96 % para Antioquia y 1,88 % para Medellín**.

Al comparar los resultados, se observa que **Colombia presenta la mayor tasa de letalidad**, seguida de Antioquia y Medellín. Esto indica que, proporcionalmente, la relación entre fallecimientos y casos confirmados fue ligeramente mayor en el ámbito nacional. Por el contrario, Medellín presenta la menor tasa de los tres niveles analizados. Estos resultados permiten comparar el comportamiento de la mortalidad por COVID-19 entre diferentes niveles geográficos, teniendo en cuenta que la tasa representa una proporción de los casos confirmados y no el riesgo individual de fallecer por la enfermedad.

¿Cuáles son los retrasos promedio entre las fechas? (5 puntos).




```python
retrasos = {
    'Síntomas → Diagnóstico': ('Fecha de inicio de síntomas', 'Fecha de diagnóstico'),
    'Síntomas → Notificación': ('Fecha de inicio de síntomas', 'Fecha de notificación'),
    'Diagnóstico → Reporte web': ('Fecha de diagnóstico', 'fecha reporte web'),
    'Síntomas → Recuperación': ('Fecha de inicio de síntomas', 'Fecha de recuperación'),
    'Síntomas → Deceso': ('Fecha de inicio de síntomas', 'Fecha de muerte')
}

for nombre, (inicio, fin) in retrasos.items():
    promedio = (df_copia[fin] - df_copia[inicio]).dt.days.mean()
    print(f"{nombre}: {promedio:.1f} días")
```

**Justificación**

El análisis de las fechas permitió identificar diferentes tiempos promedio en el proceso de atención y seguimiento de los casos de COVID-19. En promedio, transcurrieron **8,4 días entre el inicio de síntomas y el diagnóstico**, **4,2 días entre el inicio de síntomas y la notificación** y **4,4 días entre el diagnóstico y el reporte web**.

En cuanto a la evolución de los casos, el tiempo promedio entre el inicio de síntomas y la recuperación fue de **23,0 días**, mientras que entre el inicio de síntomas y el deceso fue de **44,8 días**. Estos resultados permiten observar los diferentes intervalos temporales presentes en el registro de los casos y pueden ser útiles para analizar el comportamiento y evolución de la enfermedad.
